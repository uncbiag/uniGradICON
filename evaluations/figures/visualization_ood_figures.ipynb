```python
from matplotlib.ticker import ScalarFormatter
import matplotlib.ticker as ticker

def show_exp(ax1, ax2, metric, dataset, ylabel=["",""], xlabel=""):
    # Draw right part
    df = pd.DataFrame(list(metric["%J"].items()), columns=["model",'%J'])
    df["dataset"] = dataset

    sns.barplot(data=df, y="%J", x='dataset', hue="model", alpha=0.5, ax=ax2, palette="Set2", width=0.7, gap=0.2)
    # ax2.set_yscale('log', subs=(2,3,4,5))
    # ax2.invert_yaxis()  # reverse the direction
    ax2.set_ylim(0, 1)
    ax2.invert_yaxis()  # reverse the direction
    ax2.spines['bottom'].set_visible(False)
    ax2.spines['right'].set_visible(False)
    ticks = ax2.get_yticklabels()
    ticks[0] = ""
    # ax2.set_yticklabels(ticks[1:])
    ax2.set_yticklabels(ticks)
    ax2.set_xticks([])
    ax2.set_xlabel(xlabel)

    # ax2.set_ylim(1e-5, 1e0)
    
    ax2.legend_.remove()

    # Draw left part
    df = pd.DataFrame(list(metric["accuracy"].items()), columns=["model",'Accuracy'])
    df["dataset"] = dataset

    # Plot the initial metrics
    # if "l2r_top5" in metric:
    #     ax1.axhline(y=metric["l2r_top5"]["top1"], color='black', linestyle='--', label='Initial', linewidth=0.8)
    #     ax1.axhline(y=metric["l2r_top5"]["top5"], color='black', linestyle='--', label='Initial', linewidth=0.8)

    sns.barplot(data=df, y="Accuracy", x='dataset', hue="model", alpha=0.9, ax=ax1, palette="Set2", width=0.7, gap=0.2, clip_on=False)
    # ax1.invert_yaxis()  # reverse the direction
    # ax1.tick_params(labelleft=False, left=False)
    ax1.set_xticks([])
    ax1.set_xlabel('')
    ax1.set_ylim(0, 100)
    ax1.spines['top'].set_visible(False)
    ax1.spines['right'].set_visible(False)
    ax1.legend_.remove()
    
    ax1.set_ylabel(ylabel[0])
    ax2.set_ylabel(ylabel[1])

    ax1.relim()
    ax1.autoscale_view()
    ax1.margins(y=0.01)

    ax2.relim()
    ax2.autoscale_view()
    ax2.margins(y=0.01)

     # Plot the initial metrics
    if "l2r_top5" in metric:
        ax1.axhline(y=metric["l2r_top5"]["top1"], color='salmon', linestyle='--', label='L2R Top1', linewidth=0.8)
        ax1.axhline(y=metric["l2r_top5"]["top5"], color='royalblue', linestyle='--', label='L2R Top5', linewidth=0.8)
    
    ax1.axhline(y=metric["initial"], color='black', linestyle='--', label='Initial', linewidth=0.8)
    
```

```python
import seaborn as sns
import matplotlib.gridspec as gridspec

sns.set_palette("Set1")
sns.set_theme(style='ticks')

fig = plt.figure(figsize=(20, 4))
gs = gridspec.GridSpec(2, 6, hspace=0.06, width_ratios=[1,1,1,1,1,1], height_ratios=[2, 1])

for idx, key in enumerate(list(metrics.keys())):
    # fig, (ax1, ax2) = plt.subplots(nrows=2, figsize=(3, 5), sharex=True, gridspec_kw={'hspace': 0.02, 'height_ratios':[2, 1]})
    ax1 = fig.add_subplot(gs[0, idx])
    ax2 = fig.add_subplot(gs[1, idx])
    font_size = 14

    metric = metrics[key]

    if key == "L2R-NLST":
        xlabel = f"L2R-NLST (Type {metric['type']}) \n Initial: {metric['initial']:.2f} mm"
        show_exp(ax1, ax2, metric, key, ylabel=[r"mTRE$\downarrow$ (mm)", r"$\%|J_{<0}|\downarrow$"], xlabel=xlabel)
        ax1.set_ylim(0, 6)
    elif key == "L2R-OASIS":
        xlabel = f"L2R-OASIS (Type {metric['type']}) \n Initial: {metric['initial']:.2f}"
        show_exp(ax1, ax2, metric, key, ylabel=[r"DICE$\uparrow$", ""], xlabel=xlabel)
    else:
        xlabel = f"{key} (Type {metric['type']}) \n Initial: {metric['initial']:.2f}"
        show_exp(ax1, ax2, metric, key, ylabel=["", ""], xlabel=xlabel)
    
    # if key == "IXI":
    #     ax1.legend().set_visible(True)

lines_labels = [ax.get_legend_handles_labels() for ax in fig.axes]
handles, labels = [sum(lol, []) for lol in zip(*lines_labels)]
# Remove duplicates
new_labels = {}
for i in range(len(labels)):
    if labels[i] not in new_labels:
        new_labels[labels[i]] = handles[i]
fig.legend(list(new_labels.values()), list(new_labels.keys()), loc='upper left', fontsize=font_size, ncol=7, bbox_to_anchor=(0.24, -0.1), borderaxespad=0., title="", title_fontsize=font_size)

    # sns.despine(top=False, trim=True)
plt.subplots_adjust(wspace=0.4, hspace=0.4)
plt.tight_layout()

# plt.show()
plt.savefig("./OOD_evaluations.png", dpi=300, bbox_inches='tight')
```
